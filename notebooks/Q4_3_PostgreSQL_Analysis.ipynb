```python
import sqlite3
import pandas as pd
```

```python
# Connect to the database
conn = sqlite3.connect("sanews.db")

# Import the cleaned dataset
df = pd.read_csv("sanews_cleaned.csv")

# Store the dataset in an SQLite table
df.to_sql("sanews_articles", conn, if_exists="replace", index=False)
```

```python
#articles related to education
query2 = """
SELECT
    article_url,
    article_length,
    article_text
FROM sanews_articles
WHERE article_text REGEXP '(?i)\\b(education|school|schools|university|universities|students|learners)\\b'
ORDER BY article_length DESC;
"""

pd.read_sql_query(query2, conn)
```

```python
#What other topics are covered in the SAnews articles?

query2 = """
SELECT 'Economy and employment' AS topic, COUNT(*) AS total_articles
FROM sanews_articles
WHERE article_text REGEXP '(?i)\\b(economy|economic|employment|unemployment|jobs|investment|inflation|GDP)\\b'

UNION ALL

SELECT 'Education', COUNT(*)
FROM sanews_articles
WHERE article_text REGEXP '(?i)\\b(education|school|schools|university|universities|students|learners)\\b'

UNION ALL

SELECT 'Health', COUNT(*)
FROM sanews_articles
WHERE article_text REGEXP '(?i)\\b(health|hospital|hospitals|clinic|clinics|disease|vaccination)\\b'

UNION ALL

SELECT 'Agriculture', COUNT(*)
FROM sanews_articles
WHERE article_text REGEXP '(?i)\\b(agriculture|agricultural|farming|farmers|crops)\\b'

UNION ALL

SELECT 'Crime and safety', COUNT(*)
FROM sanews_articles
WHERE article_text REGEXP '(?i)\\b(crime|criminal|police|SAPS|arrest|arrests|safety)\\b'

UNION ALL

SELECT 'Weather and environment', COUNT(*)
FROM sanews_articles
WHERE article_text REGEXP '(?i)\\b(weather|rainfall|flood|flooding|climate|environment)\\b'

ORDER BY total_articles DESC;
"""

pd.read_sql_query(query2, conn)
```
